```python
import joblib
import pandas as pd

# 1) Load the tuned model
best_xgb = joblib.load("models/best_xgb.joblib")
THRESHOLD = 0.5
```

```python
def predict_delay_from_features(row, model=best_xgb, threshold=THRESHOLD):
    """
    row: a 1-row pandas DataFrame with the same columns as X_test_adv_s
    """
    proba_delay = model.predict_proba(row)[:, 1][0]
    label = int(proba_delay >= threshold)
    return proba_delay, label
```

```python

adv = pd.read_csv("../results/processed_features_advanced.csv")
y_adv = pd.read_csv("../results/target_advanced.csv")["arr_delay"] 

for i in range(50):
    idx = adv.sample(1, random_state=i).index[0]
    row = adv.loc[[idx]]
    proba, label = predict_delay_from_features(row)
    true = y_adv.iloc[idx]   # or whatever your target name is
    print(i, f"proba={proba:.3f}", "pred=", label, "true=", true)

print(f"Predicted delay probability: {proba:.3f}")
print("Prediction:", "DELAY" if label == 1 else "ON-TIME")
```
